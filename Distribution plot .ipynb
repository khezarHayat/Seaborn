```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
#figlevel->displot
#axeslevel->histplot,kdeplot,rugplot
```

```python
#univarent histogram
tips
```

```python
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
  # bin parameter we used want to incease the bin for histogram
sns.displot(data=tips,x='total_bill',kind='hist',bins=30) 
```

```python
sns.displot(data=tips,x='total_bill',kind='hist',hue='sex',bins=30) 
```

```python
sns.displot(data=tips,x='total_bill',kind='hist',element='step',hue='sex',bins=30) 
# element parameter
```

```python
tips
```

```python
#kdeplot
sns.kdeplot(data=tips,x='total_bill')    
```

```python
sns.displot(data=tips,x='total_bill',kind='kde') 
```

```python
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex')
```

```python
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True)
```

```python
#RUGPLOT display the same information is another style
```

```python
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True)
sns.rugplot(data=tips,x='total_bill',kind='kde',hue='sex')
```
