```python
import numpy as mp
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Scatter plot kind of catagorical
```

```python
tips=sns.load_dataset('tips')
```

```python
tips
```

```python
#strip plot
sns.stripplot(data=tips,x='day',y='total_bill',color='green')# it is axes level plot
```

```python
# there are another plot which is figure level funtion capplot
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='strip',jitter=0)    # jitter we used to adjuct the points 
```

```python
#swarmplot
sns.catplot(data=tips,x='day',y='total_bill',kind='swarm',hue='sex')
```

```python
  #catagorical distribution plot
```

```python
#boxplot
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
sns.boxplot(data=tips,y='total_bill')
```

```python
#violineplot  (boxplot+KDEplot)
```

```python
sns.violinplot(data=tips,x='day',y='total_bill',hue='sex')
```
