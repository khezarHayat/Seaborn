```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips=sns.load_dataset('tips')     # this dataset in bultin in seaborn
```

```python
tips
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip')  #same like but only write data b/c seaborn understand that grpgh is make on data# scatter plot is
```

```python
# the above saccetr plot is axes base 
#relplot is figure base same to same grapgh but diff base
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size') # this all parameer are also present in scatter plot
# it is all parameter present in this also for mutivariount data we plot 4 colum
```
